````python
district_level = df4.groupby(['State', 'Year']).agg(
    District_Transactions=('Transactions', 'sum'),
    District_Amount_INR=('Amount (INR)', 'sum'),
    District_Registered_Users=('Registered Users', 'sum')
).reset_index()
district_level['District_Amount_INR'] = district_level['District_Amount_INR'].round(2)

display(district_level.head())
````

````python
df1_yearly = df1.groupby(['State', 'Year']).agg(
    Transactions=('Transactions', 'sum'),
    Amount_INR=('Amount (INR)', 'sum'),
    Registered_Users=('Registered Users', 'sum')
).reset_index()
df1_yearly['Amount_INR'] = df1_yearly['Amount_INR'].round(2)

df_merged = pd.merge(df1_yearly, district_level, on=['State', 'Year'], how='inner')
display(df_merged.head())
````

````python
df_merged['Amount_INR'] = df_merged['Amount_INR'].round(2)
df_merged['District_Amount_INR'] = df_merged['District_Amount_INR'].round(2)

df_merged['Transactions_Diff'] = abs(df_merged['Transactions'] - df_merged['District_Transactions'])
df_merged['Amount_Diff'] = abs(df_merged['Amount_INR'] - df_merged['District_Amount_INR'])
df_merged['Registered_Users_Diff'] = abs(df_merged['Registered_Users'] - df_merged['District_Registered_Users'])

discrepancies = df_merged[
    (df_merged['Transactions_Diff'] > 0) |
    (df_merged['Amount_Diff'] > 0) |
    (df_merged['Registered_Users_Diff'] > 0)
]
if not discrepancies.empty:
    print("Discrepancies found between state-level and aggregated district-level data:")
    display(discrepancies[['State', 'Year',
                           'Transactions', 'District_Transactions', 'Transactions_Diff',
                           'Amount_INR', 'District_Amount_INR', 'Amount_Diff',
                           'Registered_Users', 'District_Registered_Users', 'Registered_Users_Diff']])
else:
    print("No discrepancies found. State-level and aggregated district-level data are consistent.")
````

## Prepare Data for Time Series Plot


First, I will filter the df1 DataFrame to include only data for the 'Maharashtra' state. This is the initial step to focus on the required state data before further processing.



````python
maharashtra_data = df1[df1['State'] == 'Maharashtra'].copy()
maharashtra_data.head()
````

Now, I will create a new column named Time_Period in the maharashtra_data DataFrame by combining the 'Year' and 'Quarter' columns. This will create a single time dimension for time series analysis.



````python
maharashtra_data['Time_Period'] = maharashtra_data['Year'].astype(str) + '-Q' + maharashtra_data['Quarter'].astype(str)
maharashtra_data.head()
````


Now that the Time_Period column is created, I will group maharashtra_data DataFrame by Time_Period and calculate the sum of 'Transactions' and 'Amount (INR)' for each time period, and will store the result in a new DataFrame named maharashtra_time_series. Then, I will sort this DataFrame by Time_Period to ensure chronological order for time series analysis.



````python
maharashtra_time_series = maharashtra_data.groupby('Time_Period').agg(
    Total_Transactions=('Transactions', 'sum'),
    Total_Amount_INR=('Amount (INR)', 'sum')
).reset_index()

maharashtra_time_series['Time_Period'] = pd.Categorical(
    maharashtra_time_series['Time_Period'],
    categories=maharashtra_data['Time_Period'].unique(),
    ordered=True
)
maharashtra_time_series = maharashtra_time_series.sort_values('Time_Period')
display(maharashtra_time_series.head())
````


Now that the (maharashtra_time_series) DataFrame is prepared, I will generate a time series plot with a dual y-axis to visualize the trend of 'Total Transactions' and 'Total Amount (INR)' over 'Time_Period' for Maharashtra.



````python
plt.figure(figsize=(15, 7))

# Plot Total Transactions
ax1 = sns.lineplot(
    x='Time_Period',
    y='Total_Transactions',
    data=maharashtra_time_series,
    marker='o',
    color='blue',
    label='Total Transactions'
)
ax1.set_xlabel('Time Period')
ax1.set_ylabel('Total Transactions', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

# Create a second y-axis for Total Amount (INR)
ax2 = ax1.twinx()
sns.lineplot(
    x='Time_Period',
    y='Total_Amount_INR',
    data=maharashtra_time_series,
    marker='x',
    color='red',
    label='Total Amount (INR)',
    ax=ax2
)
ax2.set_ylabel('Total Amount (INR)', color='red')
ax2.tick_params(axis='y', labelcolor='red')

# Add title and labels
plt.title('Total Transactions and Amount in Maharashtra Over Time')
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.7)

# Combine legends
lines, labels = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines + lines2, labels + labels2, loc='upper left')

plt.tight_layout()
plt.show()
````
